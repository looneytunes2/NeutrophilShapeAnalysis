```python
from neutrophil_shape.CustomFunctions import image_processing
from neutrophil_shape.config.loader import load_config
config = load_config(microscope_type='confocal')
```

```python
############################# RUN TRACKMATE TRACKING ON SEGMENTED IMAGES ##################################
```

```python
############# SEGMENT RAW IMAGES FOR TRACKING AND POSITION INFORMATION ############
for e, ex in vars(config.experiment).items():
    if e != 'lls':
        raw_dir = ex.serverdir
        foldlist = ex.dates
        imdir = ex.localdir
        print(raw_dir,foldlist,imdir)

        if __name__ ==  '__main__':
            image_processing.segment_whole_images(
                raw_dir,  # parent directory for the folders from different imaging days
                foldlist,  # the dates on the folders from different imaging days
                imdir,  # where to save the segmented tracking images
                config,  # Config class
                )
```

```python
############### SEGMENT AND SAVE CELLS ################################
for e, ex in vars(config.experiment).items():
    if e != 'lls':
        raw_dir = ex.serverdir
        foldlist = ex.dates
        imdir = ex.localdir
        print(raw_dir,foldlist,imdir)
        if __name__ ==  '__main__':
            image_processing.segment_and_crop_confocal(
                raw_dir, #directory with original images (saved as individual slices)
                imdir, # where the segmented tracking images are saved
                config, # Config class
                )

```

```python
### get all local data directories with processed data
foldlist = [ex.localdir for e, ex in vars(config.experiment).items() if e != 'lls']

########## GET TRAJECTORIES FROM POSITION INFO
for f in foldlist:
    if __name__ ==  '__main__':
        image_processing.get_smooth_trajectories(
            f,
            config,
            )
```
